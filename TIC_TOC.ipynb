```python

import random

wins = [
    [0, 1, 2], [3, 4, 5], [6, 7, 8],
    [0, 3, 6], [1, 4, 7], [2, 5, 8],
    [0, 4, 8], [2, 4, 6]
]
```

```python
def check_winner(board):
    for a, b, c in wins:
        if board[a] == board[b] == board[c] != ' ':
            return board[a]

    if ' ' not in board:
        return 'Draw'

    return None
```

```python

def print_board(board):
    print(board[0], "|", board[1], "|", board[2])
    print("---------")
    print(board[3], "|", board[4], "|", board[5])
    print("---------")
    print(board[6], "|", board[7], "|", board[8])
```

```python
q_table = {}

alpha = 0.2
gamma = 0.9
epsilon = 0.3

# 4. Training the model
for episode in range(10000):

    board = [' '] * 9

    while True:
        state = "".join(board)

        if state not in q_table:
            q_table[state] = [0] * 9

        available = [i for i in range(9) if board[i] == ' ']

        if random.random() < epsilon:
            action = random.choice(available)
        else:
            action = max(available, key=lambda i: q_table[state][i])

        board[action] = 'X'
        result = check_winner(board)

        reward = 0

        if result == 'X':
            reward = 1
        elif result == 'Draw':
            reward = 0.5

        if result is not None:
            next_max = 0
        else:
            available = [i for i in range(9) if board[i] == ' ']
            board[random.choice(available)] = 'O'

            result = check_winner(board)

            if result == 'O':
                reward = -1
            elif result == 'Draw':
                reward = 0.5

            next_state = "".join(board)

            if next_state not in q_table:
                q_table[next_state] = [0] * 9

            next_max = 0 if result is not None else max(q_table[next_state])

        old_value = q_table[state][action]

        q_table[state][action] = old_value + alpha * (
            reward + gamma * next_max - old_value
        )

        if result is not None:
            break

print("Training completed!")
```

```python
board = [' '] * 9

while True:
    state = "".join(board)
    available = [i for i in range(9) if board[i] == ' ']

    action = max(
        available,
        key=lambda i: q_table.get(state, [0] * 9)[i]
    )

    board[action] = 'X'

    print("\nComputer (X):")
    print_board(board)

    result = check_winner(board)

    if result is not None:
        print("Game result:", result)
        break

    available = [i for i in range(9) if board[i] == ' ']

    move = int(input("Your turn (O). Enter position 0-8: "))

    if move not in available:
        print("Invalid move! Try again.")
        board[action] = ' '
        continue

    board[move] = 'O'

    print("\nYou (O):")
    print_board(board)

    result = check_winner(board)

    if result is not None:
        print("Game result:", result)
        break
```
