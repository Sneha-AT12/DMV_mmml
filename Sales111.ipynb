```python
sns.boxplot(x='SALES', data=df)
plt.title('Sales Outlier Detection')
plt.show()
```

```python
Q1 = np.percentile(df['SALES'], 25, method='midpoint')
Q3 = np.percentile(df['SALES'], 75, method='midpoint')
IQR = Q3 - Q1

print("Old Shape:", df.shape)
```

```python

upper = np.where(df['SALES'] >= (Q3 + 1.5 * IQR))[0]
lower = np.where(df['SALES'] <= (Q1 - 1.5 * IQR))[0]

df = df.drop(df.index[np.unique(np.concatenate((upper, lower)))])

print("New Shape:", df.shape)
```

```python
sns.boxplot(x='SALES', data=df)
plt.title('Sales After Outlier Removal')
plt.show()
```
